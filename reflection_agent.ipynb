```python
import os
from groq import Groq
from IPython.display import display_markdown

class ReflectionAgent:
    def __init__(self):
        """
        Initializes the agent and ensures the Groq client is set up properly.
        It expects GROQ_API_KEY to be present in the environment variables.
        """
        if not os.environ.get("GROQ_API_KEY"):
            raise ValueError("GROQ_API_KEY not found in environment. Please load your .env file or run getpass().")

        # Initialize the shared Groq client
        self.client = Groq()

    def run(self, user_msg, generation_system_prompt, reflection_system_prompt, n_steps=3, verbose=1):
        """
        Runs the autonomous Generation-Reflection loop.

        Args:
            user_msg (str): The initial task (e.g., 'Generate Merge Sort')
            generation_system_prompt (str): The persona instruction for the coder
            reflection_system_prompt (str): The persona instruction for the critic
            n_steps (int): Total number of refinement rounds to run
            verbose (int): If 1, prints out live status tracking

        Returns:
            str: The final, polished code response
        """
        # 1. Initialize the Coder's conversation thread
        generation_chat_history = [
            {"role": "system", "content": generation_system_prompt},
            {"role": "user", "content": user_msg}
        ]

        current_code = ""

        # 2. Start the Agent Loop
        for step in range(1, n_steps + 1):
            if verbose == 1:
                print(f"\n🚀 [Step {step}/{n_steps}] Coder Agent is generating/refining code...")

            # --- GENERATION PHASE ---
            # Call Groq to get the current code draft
            coder_response = self.client.chat.completions.create(
                messages=generation_chat_history,
                model="llama-3.3-70b-versatile"
            )
            current_code = coder_response.choices[0].message.content

            # Save the code to the coder's history so it stays in context
            generation_chat_history.append({"role": "assistant", "content": current_code})

            # If we reached the final step, we don't need to ask for a critique anymore
            if step == n_steps:
                break

            # --- REFLECTION PHASE ---
            if verbose == 1:
                print(f"🧐 [Step {step}/{n_steps}] Critic Agent (Andrej Karpathy) is reviewing the code...")

            # Initialize a completely fresh reviewer list for this specific round
            reflection_chat_history = [
                {"role": "system", "content": reflection_system_prompt},
                {"role": "user", "content": current_code}
            ]

            # Call Groq to get the critique from the reviewer persona
            critic_response = self.client.chat.completions.create(
                messages=reflection_chat_history,
                model="llama-3.3-70b-versatile"
            )
            critique = critic_response.choices[0].message.content

            if verbose == 1:
                print(f"📝 [Step {step}/{n_steps}] Critique received! Passing feedback back to the Coder.")

            # --- CLOSING THE LOOP ---
            # Feed the critique back into the Coder's chat history as a new user challenge
            feedback_msg = f"Here is a critique of your code from a senior engineer:\n\n{critique}\n\nPlease rewrite the code fixing these issues."
            generation_chat_history.append({"role": "user", "content": feedback_msg})

        if verbose == 1:
            print("\n✅ Agent loop completed successfully!")

        return current_code
```

```python
from getpass import getpass
import os

# Secure API step (if not already done)
if not os.environ.get("GROQ_API_KEY"):
    os.environ["GROQ_API_KEY"] = getpass("Enter your Groq API Key: ")

# 1. Instantiate your brand-new custom agent class
agent = ReflectionAgent()

# 2. Define the configuration strings
generation_system_prompt = "You are a Python programmer tasked with generating high quality Python code."
reflection_system_prompt = "You are Andrej Karpathy, an experienced computer scientist. Look for performance bottlenecks, edge-case bugs, and code cleanliness flaws."
user_msg = "Generate a Python implementation of the Merge Sort algorithm"

# 3. Run the engine (set to 3 steps so it finishes quickly while demonstrating the loop)
final_response = agent.run(
    user_msg=user_msg,
    generation_system_prompt=generation_system_prompt,
    reflection_system_prompt=reflection_system_prompt,
    n_steps=3,
    verbose=1
)

# 4. Render the masterfully refined code output
display_markdown(final_response, raw=True)
```
